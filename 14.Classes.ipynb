# Sobre classes

Em programação orientada a objetos, classes permitem organizar o código em blocos que representam entidades ou conceitos. Classes ajudam a estruturar programas de maneira lógica e modular, facilitando a criação, manutenção e expansão de projetos de software.

### Facilidade de Manutenção e Expansão

Com classes, fica mais fácil manter o código e expandir suas funcionalidades. Alterações em uma classe afetam todos os lugares onde a classe é usada, evitando a necessidade de fazer mudanças em múltiplos pontos do código. Ao adicionar novos recursos, você pode criar novas classes ou herdar de classes existentes, mantendo o código organizado e fácil de entender.

Quando seu projeto cresce, organizar o código em vários arquivos é uma boa prática para manter o código limpo, modular e fácil de manter. Em Python, isso é feito separando classes em módulos (arquivos .py), ou também agrupando, classes relacionadas em um arquivo ou diretório específico, e, em seguida, importando-as quando necessário.

├── cachorro.py   # Classe Cachorro

Aqui estão alguns dos principais motivos para usar classes:


1. Organização e Estruturação do Código

    Classes permitem agrupar dados e comportamentos (atributos e métodos) em uma única entidade. Isso facilita a compreensão e organização do código.
    Em vez de ter variáveis e funções soltas, você pode encapsular informações e métodos relacionados a uma entidade em uma única estrutura.

    No exemplo abaixo, a classe Cachorro agrupa tudo o que é relacionado a um "cachorro" em um lugar: o nome, a idade e o comportamento de "latir".

```python
class Cachorro:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def latir(self):
        print("Woof!")


meu_cachorro = Cachorro("Rex", 3)
meu_cachorro.latir()
```

# Métodos Mágicos 
Métodos “Dunder” (double underscore) são métodos especiais no Python, escritos como __método__. 

Esses métodos fornecem uma maneira de definir comportamentos específicos para operações ou funcionalidades integradas em classes Python. Ao implementar métodos dunder, você pode personalizar o comportamento de seus objetos e fazê-los funcionar perfeitamente com as construções da linguagem Python.

No exemplo acima, temos a linha:

def __init__(self, nome, idade):

Isto quer dizer que quando um objeto desta classe é criado, este método será chamado. Ele inicializa os atributos do objeto e executa qualquer configuração necessária. Métodos de classe devem trazer a referência self, porque se refe ao objeto criado, e não a classe. Nesta função, temos também outros paâmetros como nome e idade do objeto (self).



2. Reutilização e Modularidade (com Herança e Composição)

    Classes facilitam a reutilização de código. Por meio de herança, uma classe pode herdar atributos e métodos de outra, evitando duplicação de código.
    A composição permite que você combine várias classes para criar funcionalidades mais complexas.

    No exemplo abaixo, Cachorro herda de Animal, então não é necessário redefinir o método comer. Isso torna o código mais modular e reutilizável.

```python
class Animal:
    def comer(self):
        print("O animal está comendo")

class Cachorro(Animal):
    def latir(self):
        print("Woof!")

dog = Cachorro()
dog.comer()  # Saída: O animal está comendo
dog.latir()  # Saída: Woof!
```

3. Encapsulamento e Segurança

    O encapsulamento permite proteger os dados de serem acessados ou modificados diretamente. Isso ajuda a evitar erros e torna o código mais seguro.
    Com classes, é possível controlar o acesso aos atributos e métodos de um objeto, usando convenções como prefixos _ (privado) e __ (ocultação de nomes).

    Neste exemplo, __saldo é "privado" (usando __), então o saldo só pode ser acessado pelos métodos depositar e ver_saldo, o que ajuda a controlar como o saldo é manipulado

```python
class ContaBancaria:

    def __init__(self, saldo):
        self.__saldo = saldo

    def depositar(self, valor):
        self.__saldo += valor

    def ver_saldo(self):
        return self.__saldo

conta = ContaBancaria(100)
conta.depositar(50)
print(conta.ver_saldo())  # Saída: 150
```

4. Abstração e Simplicidade

    Com classes, você pode simplificar conceitos complexos e representá-los de forma abstrata.
    A classe encapsula detalhes internos, expondo apenas o que é necessário para o uso externo. Isso torna o código mais fácil de usar e entender.

    Neste exemplo, a complexidade do motor está encapsulada dentro do Carro, permitindo que o usuário do Carro apenas ligue o carro sem precisar saber como o motor funciona.

```python
class Motor:
    def iniciar(self):
        print("Motor iniciado")

class Carro:
    def __init__(self):
        self.motor = Motor()

    def ligar(self):
        self.motor.iniciar()
        print("Carro ligado")

carro = Carro()
carro.ligar()  # Saída: Motor iniciado\n Carro ligado
```

5. Polimorfismo

    Com classes, é possível usar polimorfismo, que permite que diferentes objetos respondam a uma mesma interface de maneiras distintas.
    Isso permite que uma função opere em diferentes tipos de objetos que compartilhem uma interface comum.

    Neste exemplo, fazer_o_som_do_animal funciona com qualquer objeto que implemente o método fazer_som, independentemente da classe. Isso facilita a criação de código genérico e extensível.

```python
class Pato:
    def fazer_som(self):
        print("Quack!")

class Cachorro:
    def fazer_som(self):
        print("Woof!")

def fazer_o_som_do_animal(animal):
    animal.fazer_som()

fazer_o_som_do_animal(Pato())    # Saída: Quack!
fazer_o_som_do_animal(Cachorro()) # Saída: Woof!
```
